```python

class Profesor:
    def _imprimir(self):
        print("Es un profesor.")


class ProfesorTitular(Profesor):

    def __init__(self):
        super().__init__()
        self.anos = 0

    def _imprimir(self):
        print("Es un profesor titular.")

    def _imprimir_anos(self):
        print(f"Años = {self.anos}")


class Prueba3:

    @staticmethod
    def main():
       
        profesor1: Profesor = ProfesorTitular()
        profesor1._imprimir_anos()

if __name__ == "__main__":
    Prueba3.main()
```
