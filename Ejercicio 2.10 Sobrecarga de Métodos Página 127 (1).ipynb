```python
class Pedido:
    def calcular_pedido(self, *args):
        cantidad_parametros = len(args)

        if cantidad_parametros == 4:
            primer_plato = args[0]
            costo_primer_plato = args[1]
            bebida = args[2]
            costo_bebida = args[3]
            
            total = costo_primer_plato + costo_bebida
            print(f"\n Total: El costo de {primer_plato} y {bebida} es ${total}")

        elif cantidad_parametros == 6:
            primer_plato = args[0]
            costo_primer_plato = args[1]
            segundo_plato = args[2]
            costo_segundo_plato = args[3]
            bebida = args[4]
            costo_bebida = args[5]
            
            total = costo_primer_plato + costo_segundo_plato + costo_bebida
            print(f"\n Total: El costo de {primer_plato} + {segundo_plato} + {bebida} es ${total}")


        elif cantidad_parametros == 8:
            primer_plato = args[0]
            costo_primer_plato = args[1]
            segundo_plato = args[2]
            costo_segundo_plato = args[3]
            postre = args[4]
            costo_postre = args[5]
            bebida = args[6]
            costo_bebida = args[7]
            
            total = costo_primer_plato + costo_segundo_plato + costo_bebida + costo_postre
            print(f"\n Total: El costo de {primer_plato} + {segundo_plato} + {bebida} + {postre} es ${total}")

        else:
            print("\nError: Cantidad de parámetros no válida para el pedido.")

def main():
    print("   SISTEMA DE PEDIDOS DEL RESTAURANTE   ")
    
    mi_pedido = Pedido()
    
    print("Opciones de pedido:")
    print("1. Un primer plato y una bebida")
    print("2. Un primer plato, un segundo plato y una bebida")
    print("3. Un primer plato, un segundo plato, un postre y una bebida")
    
    opcion = input("\nSeleccione el número de su opción (1, 2 o 3): ")
    
    if opcion == '1':
        print("\n--- Ingrese los datos ---")
        plato1 = input("Nombre del primer plato: ")
        costo_p1 = float(input(f"Costo de {plato1}: $"))
        bebida = input("Nombre de la bebida: ")
        costo_b = float(input(f"Costo de {bebida}: $"))
     
        mi_pedido.calcular_pedido(plato1, costo_p1, bebida, costo_b)
        
    elif opcion == '2':
        print("\n--- Ingrese los datos ---")
        plato1 = input("Nombre del primer plato: ")
        costo_p1 = float(input(f"Costo de {plato1}: $"))
        plato2 = input("Nombre del segundo plato: ")
        costo_p2 = float(input(f"Costo de {plato2}: $"))
        bebida = input("Nombre de la bebida: ")
        costo_b = float(input(f"Costo de {bebida}: $"))
        
        mi_pedido.calcular_pedido(plato1, costo_p1, plato2, costo_p2, bebida, costo_b)
        
    elif opcion == '3':
        print("\n--- Ingrese los datos ---")
        plato1 = input("Nombre del primer plato: ")
        costo_p1 = float(input(f"Costo de {plato1}: $"))
        plato2 = input("Nombre del segundo plato: ")
        costo_p2 = float(input(f"Costo de {plato2}: $"))
        postre = input("Nombre del postre: ")
        costo_po = float(input(f"Costo de {postre}: $"))
        bebida = input("Nombre de la bebida: ")
        costo_b = float(input(f"Costo de {bebida}: $"))
        
        mi_pedido.calcular_pedido(plato1, costo_p1, plato2, costo_p2, postre, costo_po, bebida, costo_b)
        
    else:
        print("\nOpción no válida. Por favor reinicie el programa.")

if __name__ == "__main__":
    main()
```
