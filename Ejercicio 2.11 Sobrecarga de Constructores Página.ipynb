```python
class ArticuloCientifico:

    def __init__(self, titulo, autor, palabras_claves=None, publicacion=None, ano=None, resumen=None):
        self.titulo = titulo
        self.autor = autor
        self.palabras_claves = palabras_claves if palabras_claves is not None else []
        self.publicacion = publicacion if publicacion is not None else "No especificada"
        self.ano = ano if ano is not None else 0
        self.resumen = resumen if resumen is not None else "Sin resumen"

    def imprimir(self):
        print(" DATOS DEL ARTÍCULO CIENTÍFICO ")
        print(f"Título del artículo = {self.titulo}")
        print(f"Autor del artículo = {self.autor}")
        print("Palabras clave = ")
        
        for palabra in self.palabras_claves:
            print(f"  > {palabra}")
            
        print(f"Publicación = {self.publicacion}")
        print(f"Año = {self.ano}")
        print(f"Resumen = {self.resumen}")
        print("-" * 50)

def main():
    print(" REGISTRO DE ARTÍCULOS CIENTÍFICOS  ")
    print("\nIngrese los datos para instanciar el artículo usando el constructor completo :\n")
    

    titulo_in = input("Ingrese el título del artículo: ")
    autor_in = input("Ingrese el autor: ")
    
    print("\nIngrese 3 palabras clave:")
    palabra1 = input(" - Palabra clave 1: ")
    palabra2 = input(" - Palabra clave 2: ")
    palabra3 = input(" - Palabra clave 3: ")

    palabras_in = [palabra1, palabra2, palabra3] 
    
    publicacion_in = input("\nIngrese el nombre de la publicación: ")
    ano_in = int(input("Ingrese el año de publicación: "))
    resumen_in = input("Ingrese el resumen del artículo: ")
    
    print("\nInstanciando el objeto en memoria...")
    
    articulo = ArticuloCientifico(
        titulo=titulo_in, 
        autor=autor_in, 
        palabras_claves=palabras_in, 
        publicacion=publicacion_in, 
        ano=ano_in, 
        resumen=resumen_in
    )
    
    articulo.imprimir()

if __name__ == "__main__":
    main()
```
